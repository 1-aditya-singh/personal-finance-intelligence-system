```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df

# Inspect dataset
df.head()
df.shape
df.columns
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
print(df.head(10))

# Q-1  How many Income and Expense transactions do we have?
df['type'].value_counts()

# Q-2 Spending Catogries
df['category'].value_counts()

df.count()
df['category'].mode()
df[df['type']=='Income'].count()
df["type"].value_counts()
df.info()
```

```python
df['date']=pd.to_datetime(df['date'])
df.info()

# seperate income and expense
expense=df[df['type']=='Expense']
income=df[df['type']=='Income']
```

```python
# totl income and total expense

total_income=income['amount'].sum()
total_expense=expense['amount'].sum()

# Total saving
saving=total_income-total_expense

# Saving rate
saving_rate=(saving/total_income)*100

print("Total Income:",total_income)
print("Total Expense:",total_expense)
print("Saving:",saving)
print("Saving Rate:",saving_rate)
```

```python
# Step 5: Category-wise spending

# 1. Select only expense transactions
expenses = df[df["type"] == "Expense"]

# 2. Calculate total spending for each category
category_spending = expenses.groupby("category")["amount"].sum().reset_index()

# 3. Sort categories from highest spending to lowest
category_spending = category_spending.sort_values(
    by="amount",
    ascending=False
)

# 4. Display result
print(category_spending)
```

```python

#  filter expenses
total_expense = expenses["amount"].sum()

# Category wise spending
category_spending["percentage"] = (
    category_spending["amount"] / total_expense
) * 100

# category spending percentage
category_spending
category_spending["percentage"] = category_spending["percentage"].round(2)

category_spending

# higest spending category
higest_spending_category=category_spending.iloc[0]['category']
higest_spending_category
```
