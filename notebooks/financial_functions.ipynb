```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
# Total Income Function
def calculate_total_income(df):
  income=df[df['type']=='Income']
  return income['amount'].sum()

total_income = calculate_total_income(df)

print("Total Income:", total_income)
```

```python
# Total expense function
def calculate_total_expense(df):
  expense=df[df['type']=='Expense']
  return expense['amount'].sum()

total_expense = calculate_total_expense(df)

print("Total Expense:", total_expense)
```

```python
# Savings Function
def calculate_savings(df):
    total_income = calculate_total_income(df)
    total_expense = calculate_total_expense(df)

    return total_income - total_expense

savings = calculate_savings(df)

print("Savings:", savings)
```

```python
# savings rate function
def calculate_savings_rate(df):
    total_income = calculate_total_income(df)
    savings = calculate_savings(df)

    return (savings / total_income) * 100

savings_rate = calculate_savings_rate(df)

print("Savings Rate:", round(savings_rate, 2), "%")
```

```python
# Average expense function
def calculate_average_expense(df):
    expenses = df[df["type"] == "Expense"]

    return expenses["amount"].mean()

average_expense = calculate_average_expense(df)

print(
    "Average Expense:",
    round(average_expense, 2)
)
```

```python
# Largest expense function
def calculate_largest_expense(df):
    expenses = df[df["type"] == "Expense"]

    return expenses["amount"].max()

largest_expense = calculate_largest_expense(df)

print("Largest Expense:", largest_expense)
```

```python
# Financial Summary Function
def financial_summary(df):

    total_income = calculate_total_income(df)
    total_expense = calculate_total_expense(df)
    savings = calculate_savings(df)
    savings_rate = calculate_savings_rate(df)
    average_expense = calculate_average_expense(df)
    largest_expense = calculate_largest_expense(df)

    return {
        "total_income": total_income,
        "total_expense": total_expense,
        "savings": savings,
        "savings_rate": round(savings_rate, 2),
        "average_expense": round(average_expense, 2),
        "largest_expense": largest_expense
    }
```

```python
# Test Financial Summary Function
summary = financial_summary(df)

print(summary)
```

```python
print("Financial Summary")
print("-----------------")

print(f"Total Income: ₹{summary['total_income']}")
print(f"Total Expense: ₹{summary['total_expense']}")
print(f"Savings: ₹{summary['savings']}")
print(f"Savings Rate: {summary['savings_rate']}%")
print(f"Average Expense: ₹{summary['average_expense']}")
print(f"Largest Expense: ₹{summary['largest_expense']}")
```
