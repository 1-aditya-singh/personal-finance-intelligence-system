```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
#  seperate income and expense
income = df[df["type"] == "Income"]
expenses = df[df["type"] == "Expense"]
```

```python
#  basic financial metrices
total_income = income["amount"].sum()

total_expense = expenses["amount"].sum()

savings = total_income - total_expense

savings_rate = (savings / total_income) * 100

average_expense = expenses["amount"].mean()

largest_expense = expenses["amount"].max()
```

```python
# category spending
category_spending = (
    expenses.groupby("category")['amount']
    .sum()
    .sort_values(ascending=False)
)
print(category_spending)

highest_spending_category = category_spending.index[0]

highest_spending_amount = category_spending.iloc[0]
```

```python
# saving insight
if savings_rate >= 50:
    savings_insight = (
        f"💰 Excellent savings rate of "
        f"{round(savings_rate, 2)}%."
    )

elif savings_rate >= 30:
    savings_insight = (
        f"🟢 Good savings rate of "
        f"{round(savings_rate, 2)}%."
    )

elif savings_rate >= 20:
    savings_insight = (
        f"🟡 Moderate savings rate of "
        f"{round(savings_rate, 2)}%."
    )

else:
    savings_insight = (
        f"🔴 Low savings rate of "
        f"{round(savings_rate, 2)}%."
    )

print(savings_insight)
```

```python
# general spending insight
spending_insight = (
    f"🛍️ Highest spending category is "
    f"{highest_spending_category} "
    f"with ₹{highest_spending_amount} spent."
)
print(spending_insight)
```

```python
# percentage of total spending
highest_category_percentage = (
    highest_spending_amount / total_expense
) * 100

highest_category_percentage = round(
    highest_category_percentage,
    2
)

print(
    f"{highest_spending_category} accounts for "
    f"{highest_category_percentage}% of total spending."
)
```

```python
# budget insight
budgets = {
    "Food": 5000,
    "Shopping": 7000,
    "Travel": 5000,
    "Bills": 4000,
    "Education": 3000,
    "Entertainment": 3000,
    "Health": 2500
}

budget_insights = []

for category, budget in budgets.items():

    actual = category_spending.get(category, 0)

    remaining = budget - actual

    if remaining < 0:

        budget_insights.append(
            f"🔴 {category} exceeded its budget by "
            f"₹{abs(remaining)}."
        )

    elif actual >= budget * 0.9:

        budget_insights.append(
            f"🟠 {category} is using "
            f"{round((actual / budget) * 100, 2)}% "
            f"of its budget."
        )


print("Budget Insights")
print("---------------")

for insight in budget_insights:
    print(insight)
```

```python
# Large Expense Insight
if largest_expense >= 1500:

    large_expense_insight = (
        f"⚠️ Your largest single expense was "
        f"₹{largest_expense}."
    )

else:

    large_expense_insight = (
        f"✅ Your largest single expense was "
        f"₹{largest_expense}."
    )

print(large_expense_insight)
```

```python
# Monthly Performance
# Ensure 'date' column is datetime
df["date"] = pd.to_datetime(df["date"])
# Create 'month' column
df["month"] = df["date"].dt.to_period("M")

# Calculate monthly income and expense using df directly
monthly_income = (
    df[df["type"] == "Income"]
    .groupby("month")["amount"]
    .sum()
)

monthly_expense = (
    df[df["type"] == "Expense"]
    .groupby("month")["amount"]
    .sum()
)

monthly_summary = pd.DataFrame({
    "income": monthly_income,
    "expense": monthly_expense
})

# Fill NaN values with 0, in case a month has no income or no expense
monthly_summary = monthly_summary.fillna(0)

monthly_summary["savings"] = (
    monthly_summary["income"]
    - monthly_summary["expense"]
)

# Calculate savings rate for monthly_summary
# Handle potential division by zero if income is 0 for a month
monthly_summary["savings_rate"] = (
    monthly_summary["savings"] / monthly_summary["income"].replace(0, pd.NA)
) * 100
# Fill NaN values (where income was 0) with a default (e.g., 0 or some indicator)
monthly_summary["savings_rate"] = monthly_summary["savings_rate"].fillna(0)

# find best month
best_month = monthly_summary["savings_rate"].idxmax()

best_month_rate = monthly_summary.loc[
    best_month,
    "savings_rate"
]

print(
    f"📈 Best financial month was {best_month} "
    f"with a savings rate of {best_month_rate:.2f}%."
)

# find weakest month
worst_month = monthly_summary["savings_rate"].idxmin()

worst_month_rate = monthly_summary.loc[
    worst_month,
    "savings_rate"
]


print(
    f"📉 Lowest savings rate occurred in "
    f"{worst_month} at {worst_month_rate:.2f}%."
)
```

```python
# final insight
print("===================================")
print("   PERSONAL FINANCE INSIGHTS")
print("===================================")

print()

print("💰 SAVINGS")
print(savings_insight)

print()

print("🛍️ SPENDING")
print(spending_insight)

print(
    f"{highest_spending_category} accounts for "
    f"{highest_category_percentage}% "
    f"of total spending."
)

print()

print("📊 BUDGET")
for insight in budget_insights:
    print(insight)

print()

print("⚠️ LARGE EXPENSE")
print(large_expense_insight)

print()

print("📈 MONTHLY PERFORMANCE")
print(
    f"Best month: {best_month} "
    f"({best_month_rate}% savings rate)"
)

print(
    f"Lowest month: {worst_month} "
    f"({worst_month_rate}% savings rate)"
)
```
