```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
# select expenses
expenses = df[df["type"] == "Expense"].copy()
```

```python
# create category budget
budgets = {
    "Food": 5000,
    "Shopping": 7000,
    "Travel": 5000,
    "Bills": 4000,
    "Education": 3000,
    "Entertainment": 3000,
    "Health": 2500
}
```

```python
# Calculate actual spending
category_spending = expenses.groupby("category")['amount'].sum()
category_spending
```

```python
# Create a budget table

budget_data = []

for category, budget in budgets.items():

    actual = category_spending.get(category, 0)

    remaining = budget - actual

    budget_data.append({
        "category": category,
        "budget": budget,
        "actual_spending": actual,
        "remaining": remaining
    })

budget_summary = pd.DataFrame(budget_data)

budget_summary
```

```python
# Add budget status
budget_summary["status"] = budget_summary["remaining"].apply(
    lambda x: "Over Budget" if x < 0 else "Within Budget"
)
budget_summary
```

```python
# Calculate budget utilization
budget_summary["utilization"] = (
    budget_summary["actual_spending"]
    / budget_summary["budget"]
) * 100

#  rount it
budget_summary["utilization"] = (
    budget_summary["utilization"].round(2)
)
```

```python
# Find over-budget categories
over_budget = budget_summary[
    budget_summary["remaining"] < 0
]

print("Over Budget Categories")
print("----------------------")

print(over_budget)
```

```python
print("Budget Tracking Report")
print("----------------------")

for index, row in budget_summary.iterrows():

    print(
        f"{row['category']}: "
        f"Budget ₹{row['budget']} | "
        f"Spent ₹{row['actual_spending']} | "
        f"Used {row['utilization']}% | "
        f"{row['status']}"
    )
```
