```python
#  Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
#  Convert date column
df['date']=pd.to_datetime(df['date'])
df.info()
```

```python
# Select expense
expense=df[df['type']=='Expense']
expense.head()
```

```python
# Total expense transactions
total_expense_transactions=len(expense)
print("Total Expense Transactions:", total_expense_transactions)
```

```python
# Average expense transaction
average_expense=expense['amount'].mean()
print("Average Expense Transaction:", round(average_expense, 2))
```

```python
# Find the largest single expense
largest_expense=expense['amount'].max()
print("Largest Single Expense:", largest_expense)
```

```python
# Find the smallest expense
smallest_expense=expense['amount'].min()
print("Smallest Expense:", smallest_expense)
```

```python
# Calculate median expense
median_expense = expense["amount"].median()

print("Median Expense:", median_expense)
```

```python
# find the most frequent expense category
category_transaction_count = (
    expense["category"]
    .value_counts()
)

print(category_transaction_count)
```

```python
# Find the most frequent category
expense.head()
most_frequent_category = category_transaction_count.idxmax()

print("Most Frequent Expense Category:", most_frequent_category)
```

```python
# Find the category with the highest total spending
category_spending = (
    expense.groupby("category")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_spending)
```

```python
print("Spending Behavior Summary")
print("-------------------------")

print(
    f"Total expense transactions: "
    f"{total_expense_transactions}"
)

print(
    f"Average expense per transaction: "
    f"₹{round(average_expense, 2)}"
)

print(
    f"Median expense: "
    f"₹{median_expense}"
)

print(
    f"Largest single expense: "
    f"₹{largest_expense}"
)

print(
    f"Smallest expense: "
    f"₹{smallest_expense}"
)

print(
    f"Most frequent category: "
    f"{most_frequent_category}"
)

print(
    f"Highest spending category: "
    f"{category_spending.index[0]}"
)
```
