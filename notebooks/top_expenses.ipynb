```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
# convert date
df['date']=pd.to_datetime(df['date'])
```

```python
# selecte expense
expense=df[df['type']=='Expense']
expense.head()
```

```python
# Sort expenses from largest to smallest
top_expenses=expense.sort_values(by='amount',ascending=False)
top_expenses.head(10)
```

```python
# Display Top 5 expenses
top_5_expenses = top_expenses.head(5)

top_5_expenses
```

```python
# Calculate the total of Top 5 expenses
top_5_total = top_5_expenses["amount"].sum()

print("Top 5 Expenses Total:", top_5_total)
```

```python
# Compare Top 5 with total expenses
total_expenses = expense["amount"].sum()

print("Total Expenses:", total_expenses)

top_5_percentage = (
    top_5_total / total_expenses
) * 100

print(
    "Top 5 Expenses Percentage:",
    round(top_5_percentage, 2),
    "%"
)
```

```python
# Find the largest expense category among Top 5
top_5_categories = top_5_expenses["category"].value_counts()

print(top_5_categories)

most_common_top_category = top_5_categories.idxmax()

print(
    "Most Common Category in Top 5:",
    most_common_top_category
)
```

```python
# Find the largest expense transaction
largest_transaction = top_expenses.iloc[0]

print("Largest Expense Transaction")
print("---------------------------")
print(largest_transaction)
```

```python
# Create a clean Top 5 table
top_5_clean = top_5_expenses[
    ["date", "category", "amount", "description"]
]

top_5_clean
```

```python
print("Top Expense Insights")
print("--------------------")

print(
    f"Largest single expense: "
    f"₹{largest_transaction['amount']}"
)

print(
    f"Category: "
    f"{largest_transaction['category']}"
)

print(
    f"Date: "
    f"{largest_transaction['date'].date()}"
)

print(
    f"Top 5 expenses total: "
    f"₹{top_5_total}"
)

print(
    f"Top 5 expenses represent: "
    f"{round(top_5_percentage, 2)}% of total spending"
)

print(
    f"Most common category among Top 5: "
    f"{most_common_top_category}"
)
```
