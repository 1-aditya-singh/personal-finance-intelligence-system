```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence-system/data/transactions.csv")
df.head()
```

```python
# separate income and expense
income = df[df["type"] == "Income"]
expenses = df[df["type"] == "Expense"]
```

```python
# total income
total_income = income["amount"].sum()

print("Total Income:", total_income)
```

```python
# total expense
total_expense = expenses["amount"].sum()

print("Total Expense:", total_expense)
```

```python
# total saving
savings = total_income - total_expense

print("Total Savings:", savings)
```

```python
# Calculate overall savings rate
savings_rate = (savings / total_income) * 100

print("Savings Rate:", round(savings_rate, 2), "%")
```

```python
# Monthly Income and Expense
df["date"] = pd.to_datetime(df["date"])
df["month"] = df["date"].dt.to_period("M")

# monthly income
monthly_income = df[df["type"]=='Income'].groupby("month")['amount'].sum()

# monthly expense
monthly_expense = df[df["type"]=='Expense'].groupby("month")['amount'].sum()

```

```python
# monthly saving table
monthly_savings = pd.DataFrame({
    "income": monthly_income,
    "expense": monthly_expense
})
monthly_savings
```

```python
# monthly saving
monthly_savings["savings"] = (
    monthly_savings["income"]
    - monthly_savings["expense"]
)
```

```python
# monthly saving rate
monthly_savings["savings_rate"] = (
    monthly_savings["savings"]
    / monthly_savings["income"]
) * 100
```

```python
# best saving month
best_saving_month = monthly_savings["savings"].idxmax()

print("Best Saving Month:", best_saving_month)

# to get actual amount
best_saving_amount = monthly_savings.loc[
    best_saving_month,
    "savings"
]

print("Best Saving Amount: ₹", best_saving_amount)
```

```python
#  lowest saving month
lowest_saving_month = monthly_savings["savings"].idxmin()

print("Lowest Saving Month:", lowest_saving_month)

lowest_saving_amount = monthly_savings.loc[
    lowest_saving_month,
    "savings"
]

print("Lowest Saving Amount: ₹", lowest_saving_amount)
```

```python
# average monthly saving
average_monthly_savings = monthly_savings["savings"].mean()

print(
    "Average Monthly Savings: ₹",
    round(average_monthly_savings, 2)
)
```

```python
# average saving rate
average_savings_rate = monthly_savings["savings_rate"].mean()

print(
    "Average Monthly Savings Rate:",
    round(average_savings_rate, 2),
    "%"
)
```

```python
print("Savings Analysis Report")
print("-----------------------")

print(f"Total Income: ₹{total_income}")
print(f"Total Expense: ₹{total_expense}")
print(f"Total Savings: ₹{savings}")
print(f"Overall Savings Rate: {round(savings_rate, 2)}%")

print(
    f"Average Monthly Savings: "
    f"₹{round(average_monthly_savings, 2)}"
)

print(
    f"Average Monthly Savings Rate: "
    f"{round(average_savings_rate, 2)}%"
)

print(
    f"Best Saving Month: "
    f"{best_saving_month}"
)

print(
    f"Best Saving Amount: "
    f"₹{best_saving_amount}"
)

print(
    f"Lowest Saving Month: "
    f"{lowest_saving_month}"
)

print(
    f"Lowest Saving Amount: "
    f"₹{lowest_saving_amount}"
)
```

```python
best_rate_month = monthly_savings["savings_rate"].idxmax()

best_rate = monthly_savings.loc[
    best_rate_month,
    "savings_rate"
]

print(
    f"Best savings rate was in {best_rate_month} "
    f"at {best_rate}%."
)

difference = best_saving_amount - lowest_saving_amount

print(
    f"Savings difference between best and lowest month: "
    f"₹{difference}"
)
```
