```python
# Load dataset

!git clone https://github.com/1-aditya-singh/personal-finance-intelligence-system.git

import pandas as pd
df = pd.read_csv("personal-finance-intelligence-system/data/transactions.csv")
```

```python
# Convert date column to datetime

df['date'].head()
df.dtypes
df['date']=df['date'].astype('datetime64[ns]')
df.dtypes
```

```python
# Extract month

df["month"] = df["date"].dt.to_period("M")
df[["date", "month"]].head()
```

```python
# calculate monthely income
print(df.head())
income=df[df['type']=='Income']
monthly_income=income.groupby('month')['amount'].sum()
monthly_income
```

```python
# Monthly expenses

expense = df[df["type"] == "Expense"]

monthly_expense = expense.groupby("month")["amount"].sum()

monthly_expense
```

```python
# Create monthly summary

monthly_summary = pd.DataFrame({
    "income": monthly_income,
    "expense": monthly_expense
})
monthly_summary
```

```python
# Calculate savings

monthly_summary["savings"] = (
    monthly_summary["income"]
    - monthly_summary["expense"]
)
monthly_summary
```

```python
# Calculate savings rate

monthly_summary["savings_rate"] = (
    monthly_summary["savings"]
    / monthly_summary["income"]
) * 100
monthly_summary
```

```python
# Round savings rate

monthly_summary["savings_rate"] = (
    monthly_summary["savings_rate"].round(2)
)
```

```python
# Display result
monthly_summary
```

```python
# month with highest expenses

highest_expense_month = monthly_summary["expense"].idxmax()

print("Highest Expense Month:", highest_expense_month)
print("Expense:", monthly_summary.loc[highest_expense_month, "expense"]).
```

```python
# month with higest savings
highest_saving_month = monthly_summary["savings"].idxmax()

print("Highest Saving Month:", highest_saving_month)
print("Savings:", monthly_summary.loc[highest_saving_month, "savings"])
```

```python
# highest saving rates
best_saving_rate_month = monthly_summary["savings_rate"].idxmax()

print("Best Savings Rate Month:", best_saving_rate_month)
print(
    "Savings Rate:",
    monthly_summary.loc[best_saving_rate_month, "savings_rate"],
    "%"
)
```

```python
# Financial Insights

highest_expense_month = monthly_summary["expense"].idxmax()
highest_saving_month = monthly_summary["savings"].idxmax()
best_saving_rate_month = monthly_summary["savings_rate"].idxmax()

print("Financial Insights")
print("------------------")

print(
    f"Highest spending month: {highest_expense_month}"
)

print(
    f"Highest savings month: {highest_saving_month}"
)

print(
    f"Best savings rate: {best_saving_rate_month}"
)
```

```python
# Average monthly expense
average_monthly_expense = monthly_summary["expense"].mean()

print(
    "Average Monthly Expense:",
    round(average_monthly_expense, 2)
)
```
