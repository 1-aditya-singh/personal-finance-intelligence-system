```python
# Load dataset

import pandas as pd
df = pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
# convert date column
df['date']=pd.to_datetime(df['date'])
df.head()
```

```python
# selecting expense
expense=df[df['type']=="Expense"]
expense.head()
```

```python
# Calculate spending for each day
spending_per_day = expense.groupby('date')['amount'].sum().reset_index()
spending_per_day.head()
```

```python
# Rename the column
daily_expenses = spending_per_day.rename(columns={"amount": "daily_expense"})
daily_expenses.head()
```

```python
# Find the highest spending day

highest_spending_day = daily_expenses["daily_expense"].idxmax()

print("Highest Spending Day:")
print(daily_expenses.loc[highest_spending_day])
```

```python
#  Average Daily Spending
avg_daily_spending = daily_expenses['daily_expense'].mean()
print("Average Daily Spending:", round(avg_daily_spending),2)
```

```python
# Transactions per day
daily_transactions = (
    expense.groupby("date")
    .size()
    .reset_index(name="transaction_count")
)

daily_transactions.head()
```

```python
# combine both analysis
daily_summary = pd.merge(
    daily_expenses,
    daily_transactions,
    on="date"
)

daily_summary.head()
```

```python
#  short by higest spending
daily_summary = daily_summary.sort_values(
    by="daily_expense",
    ascending=False
)

daily_summary.head(10)
```

```python
highest_spending_day = daily_summary.iloc[0]

print("Daily Spending Insights")
print("-----------------------")

print(
    f"Highest spending day: "
    f"{highest_spending_day['date'].date()}"
)

print(
    f"Amount spent: "
    f"₹{highest_spending_day['daily_expense']}"
)

print(
    f"Transactions on that day: "
    f"{highest_spending_day['transaction_count']}"
)

print(
    f"Average daily spending: "
    f"₹{round(avg_daily_spending, 2)}"
)
```
