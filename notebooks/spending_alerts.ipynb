```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence/data/transactions.csv")
df.head()
```

```python
# Select expenses
expenses=df[df['type']=='Expense'].copy()
expenses.head()
```

```python
#  high expense alert
average_expense = expenses["amount"].mean()

high_expense_limit = average_expense * 2

print("Average Expense:", round(average_expense, 2))
print("High Expense Limit:", round(high_expense_limit, 2))
```

```python
# Find high expenses
high_expenses = expenses[
    expenses["amount"] > high_expense_limit
]
high_expenses[
    ["date", "category", "amount", "description"]
]
```

```python
# High Category Spending
category_spending = expenses.groupby('category')['amount'].sum().sort_values(ascending=False)
category_spending
```

```python
# Set a category alert threshold
category_alert_limit = 5000
high_spending_categories = category_spending[
    category_spending > category_alert_limit
]
print("High Spending Categories")
print("------------------------")

print(high_spending_categories)
```

```python
# Frequent Spending
category_transaction_count = (
    expenses["category"]
    .value_counts()
)

category_transaction_count

transaction_alert_limit = 10
frequent_categories = category_transaction_count[
    category_transaction_count > transaction_alert_limit
]

print("Frequent Spending Categories")
print("----------------------------")

print(frequent_categories)
```

```python
print("Financial Spending Alerts")
print("-------------------------")

if len(high_expenses) > 0:
    print(
        f"🔴 High Expense Alert: "
        f"{len(high_expenses)} transaction(s) above "
        f"₹{round(high_expense_limit, 2)}"
    )
else:
    print("✅ No unusually high expenses detected.")


if len(high_spending_categories) > 0:
    print(
        f"🟠 Category Alert: "
        f"{len(high_spending_categories)} category/categories "
        f"above ₹{category_alert_limit}"
    )
else:
    print("✅ No high-spending categories detected.")


if len(frequent_categories) > 0:
    print(
        f"🟡 Frequency Alert: "
        f"{len(frequent_categories)} category/categories "
        f"with more than {transaction_alert_limit} transactions"
    )
else:
    print("✅ No unusually frequent spending categories detected.")
```
