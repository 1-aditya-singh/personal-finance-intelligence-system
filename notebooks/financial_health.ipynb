```python
# Load dataset
import pandas as pd
df=pd.read_csv("personal-finance-intelligence-system/data/transactions.csv")
df.head()
```

```python
total_income = df[df["type"] == "Income"]["amount"].sum()

total_expense = df[df["type"] == "Expense"]["amount"].sum()

savings = total_income - total_expense

savings_rate = (savings / total_income) * 100

average_expense = (
    df[df["type"] == "Expense"]["amount"].mean()
)

largest_expense = (
    df[df["type"] == "Expense"]["amount"].max()
)
```

```python
score = 0
if savings_rate >= 50:
    score += 50

elif savings_rate >= 30:
    score += 40

elif savings_rate >= 20:
    score += 30

elif savings_rate >= 10:
    score += 20

else:
    score += 10
```

```python
# Expense scoring
if average_expense <= 500:
    score += 30

elif average_expense <= 1000:
    score += 20

else:
    score += 10
```

```python
# Largest expense scoring
if largest_expense <= 1000:
    score += 20

elif largest_expense <= 2000:
    score += 15

else:
    score += 10
```

```python
# Determine financial health
if score >= 80:
    health_status = "Excellent"

elif score >= 60:
    health_status = "Good"

elif score >= 40:
    health_status = "Needs Improvement"

else:
    health_status = "Poor"
```

```python
print("Financial Health Assessment")
print("--------------------------")

print(f"Total Income: ₹{total_income}")
print(f"Total Expense: ₹{total_expense}")
print(f"Savings: ₹{savings}")
print(f"Savings Rate: {round(savings_rate, 2)}%")
print(f"Average Expense: ₹{round(average_expense, 2)}")
print(f"Largest Expense: ₹{largest_expense}")

print()
print(f"Financial Health Score: {score}/100")
print(f"Financial Health: {health_status}")
```
